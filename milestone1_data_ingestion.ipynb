```python
import pandas as pd

# Define the base path for your raw datasets after KaggleHub download
base_raw_path = "/kaggle/input/brazilian-ecommerce/"

customers = pd.read_csv(base_raw_path + "olist_customers_dataset.csv")
orders = pd.read_csv(base_raw_path + "olist_orders_dataset.csv")
order_items = pd.read_csv(base_raw_path + "olist_order_items_dataset.csv")
products = pd.read_csv(base_raw_path + "olist_products_dataset.csv")
sellers = pd.read_csv(base_raw_path + "olist_sellers_dataset.csv")
payments = pd.read_csv(base_raw_path + "olist_order_payments_dataset.csv")
product_category_name_translation = pd.read_csv(base_raw_path + "product_category_name_translation.csv")

print("All datasets loaded successfully into pandas DataFrames.")
```

```python
print(customers.head())
print(orders.info())
print(order_items.describe())
```

## Data Cleaning

```python
customers.drop_duplicates(inplace=True)

products['product_category_name'].fillna(
    'Unknown',
    inplace=True
)

print("Missing values in 'product_category_name' filled with 'Unknown'.")

customers.drop_duplicates(inplace=True)
orders.drop_duplicates(inplace=True)

orders['order_purchase_timestamp'] = pd.to_datetime(
    orders['order_purchase_timestamp']
)
print("Converted 'order_purchase_timestamp' to datetime objects.")
```

```python
products.isnull().sum()
```

#### Select Important Columns

From the customers, orders, products, and sellers DataFrames, select only the columns that are considered important for the subsequent analysis and create new, cleaned DataFrames (customers_clean, orders_clean, products_clean, sellers_clean).

```python
customers_clean = customers[[
    "customer_id",
    "customer_unique_id",
    "customer_city",
    "customer_state"
]]
print("Created 'customers_clean' DataFrame with selected columns.")
```

```python
orders_clean = orders[[
    "order_id",
    "customer_id",
    "order_status",
    "order_purchase_timestamp"
]]
print("Created 'orders_clean' DataFrame with selected columns.")
```

```python
products_clean = products[[
    "product_id",
    "product_category_name"
]]
print("Created 'products_clean' DataFrame with selected columns.")
```

```python
sellers_clean = sellers[[
    "seller_id",
    "seller_city",
    "seller_state"
]]
print("Created 'sellers_clean' DataFrame with selected columns.")
```
