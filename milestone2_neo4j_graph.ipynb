Step 2 — Load Orders


```python
orders_list = orders_clean.to_dict("records")

query = """
UNWIND $rows AS row
MERGE (o:Order {order_id: row.order_id})
SET o.order_status = row.order_status,
    o.order_purchase_timestamp = row.order_purchase_timestamp
"""

batch_size = 1000

with driver.session() as session:
    for i in range(0, len(orders_list), batch_size):
        batch = orders_list[i:i+batch_size]
        session.run(query, rows=batch)
        print(f"Loaded batch {i} to {i+len(batch)}")

print("All orders loaded successfully!")
```

Step 3 — Load Sellers

```python
sellers_list = sellers_clean.sample(n=10, random_state=42).to_dict("records") # Drastically reduce sample to avoid node limit issues

query = """
UNWIND $rows AS row
MERGE (s:Seller {seller_id: row.seller_id})
SET s.seller_city = row.seller_city,
    s.seller_state = row.seller_state
"""

batch_size = 2000

with driver.session() as session:
    for i in range(0, len(sellers_list), batch_size):
        batch = sellers_list[i:i+batch_size]
        session.run(query, rows=batch)

print("Sellers loaded successfully!")
```

Step 4 — Order → Product Relationship

```python
rel_list = products_clean.to_dict("records")

query = """
UNWIND $rows AS row
MATCH (o:Order {order_id: row.order_id})
MATCH (p:Product {product_id: row.product_id})
MERGE (o)-[:CONTAINS_PRODUCT]->(p)
"""

batch_size = 2000

with driver.session() as session:
    for i in range(0, len(rel_list), batch_size):
        batch = rel_list[i:i+batch_size]
        session.run(query, rows=batch)

print("Relationships loaded successfully!")
```

# Verify Graph Structure and Data Integrity
Perform Cypher queries to verify the graph structure, node counts, and relationship integrity in the Neo4j database.

```python
with driver.session() as session:
    customer_count = session.run("MATCH (c:Customer) RETURN count(c) AS count").single()["count"]
    order_count = session.run("MATCH (o:Order) RETURN count(o) AS count").single()["count"]
    product_count = session.run("MATCH (p:Product) RETURN count(p) AS count").single()["count"]
    seller_count = session.run("MATCH (s:Seller) RETURN count(s) AS count").single()["count"]

print(f"Node Counts:\nCustomers: {customer_count}\nOrders: {order_count}\nProducts: {product_count}\nSellers: {seller_count}")
```

```python
rel_seller_product_sampled = seller_product_rel[
    seller_product_rel['seller_id'].isin([s['seller_id'] for s in sellers_list]) &
    seller_product_rel['product_id'].isin(products_clean['product_id'])
]
rel_seller_product_list = rel_seller_product_sampled.to_dict("records")

query = """
UNWIND $rows AS row
MATCH (s:Seller {seller_id: row.seller_id})
MATCH (p:Product {product_id: row.product_id})
MERGE (s)-[:SELLS_PRODUCT]->(p)
"""

batch_size = 2000

with driver.session() as session:
    for i in range(0, len(rel_seller_product_list), batch_size):
        batch = rel_seller_product_list[i:i+batch_size]
        session.run(query, rows=batch)

print("Seller-Product relationships created successfully!")
```

```python
with driver.session() as session:
    has_order_rel_count = session.run("MATCH ()-[r:HAS_ORDER]->() RETURN count(r) AS count").single()["count"]
    contains_product_rel_count = session.run("MATCH ()-[r:CONTAINS_PRODUCT]->() RETURN count(r) AS count").single()["count"]
    sells_product_rel_count = session.run("MATCH ()-[r:SELLS_PRODUCT]->() RETURN count(r) AS count").single()["count"]

print(f"Relationship Counts:\nHAS_ORDER: {has_order_rel_count}\nCONTAINS_PRODUCT: {contains_product_rel_count}\nSELLS_PRODUCT: {sells_product_rel_count}")
```

```python
import pandas as pd

# Assuming base_path is defined from a previous cell
# For robust execution within this single cell, if base_path wasn't guaranteed, we'd define it here.
# base_path = "/content/drive/MyDrive/AI_KG_Project/data/processed/" # Uncomment if base_path is not globally available

# --- Create Nodes DataFrame ---

all_nodes_data = []

# Add Customer nodes
if 'customers_clean' in globals() and not customers_clean.empty:
    customer_nodes = customers_clean[['customer_id']].copy()
    customer_nodes.columns = ['id']
    customer_nodes['label'] = 'Customer'
    all_nodes_data.append(customer_nodes)

# Add Order nodes
if 'orders_clean' in globals() and not orders_clean.empty:
    order_nodes = orders_clean[['order_id']].copy()
    order_nodes.columns = ['id']
    order_nodes['label'] = 'Order'
    all_nodes_data.append(order_nodes)

# Add Product nodes
if 'products_clean' in globals() and not products_clean.empty:
    product_nodes = products_clean[['product_id']].copy()
    product_nodes.columns = ['id']
    product_nodes['label'] = 'Product'
    all_nodes_data.append(product_nodes)

# Add Seller nodes
if 'sellers_clean' in globals() and not sellers_clean.empty:
    seller_nodes = sellers_clean[['seller_id']].copy()
    seller_nodes.columns = ['id']
    seller_nodes['label'] = 'Seller'
    all_nodes_data.append(seller_nodes)

# Concatenate all node dataframes and drop duplicates
if all_nodes_data:
    graph_nodes = pd.concat(all_nodes_data, ignore_index=True).drop_duplicates(subset=['id']).reset_index(drop=True)
else:
    graph_nodes = pd.DataFrame(columns=['id', 'label'])
    print("Warning: No source dataframes found to create graph_nodes. graph_nodes will be empty.")

# --- Create Edges DataFrame ---

all_edges_data = []

# Add Customer-Order relationships
if 'customer_order_rel' in globals() and not customer_order_rel.empty:
    customer_order_edges = customer_order_rel.copy()
    customer_order_edges.columns = ['source', 'target'] # Assuming customer_id -> order_id
    customer_order_edges['type'] = 'PLACED_ORDER'
    all_edges_data.append(customer_order_edges)

# Add Order-Product relationships
if 'order_product_rel' in globals() and not order_product_rel.empty:
    order_product_edges = order_product_rel.copy()
    order_product_edges.columns = ['source', 'target'] # Assuming order_id -> product_id
    order_product_edges['type'] = 'CONTAINS_PRODUCT'
    all_edges_data.append(order_product_edges)

# Add Seller-Product relationships
if 'seller_product_rel' in globals() and not seller_product_rel.empty:
    seller_product_edges = seller_product_rel.copy()
    seller_product_edges.columns = ['source', 'target'] # Assuming seller_id -> product_id
    seller_product_edges['type'] = 'SELLS_PRODUCT'
    all_edges_data.append(seller_product_edges)

# Concatenate all edge dataframes
if all_edges_data:
    graph_edges = pd.concat(all_edges_data, ignore_index=True)
else:
    graph_edges = pd.DataFrame(columns=['source', 'target', 'type'])
    print("Warning: No source dataframes found to create graph_edges. graph_edges will be empty.")

# --- Save to CSV ---

# Ensure the directory exists if not already handled in previous cells
if not os.path.exists(base_path):
    os.makedirs(base_path, exist_ok=True)

graph_nodes.to_csv(base_path + "graph_nodes.csv", index=False)
graph_edges.to_csv(base_path + "graph_edges.csv", index=False)

print("Graph nodes and edges saved to CSV successfully!")
```
